# groupby

# 1. group on region based in store dataframe

```python
group = store.groupby("Region")

print(group)
```

# 2. Find region-wise average sales 

```python
group = store.groupby("Region")["Sales"].mean()

print(group)
```

# 3. Find region-wise average profit

```python
group = store.groupby("Region")["Profit"].mean()

print(group)
```

# 4. Find region-wise maximum profit

```python
group = store.groupby("Region")["Profit"].max()

print(group)
```

# 5. Apply multiple aggregate functions on region-wise sales

```python
def first_mode(series):
    modes = series.mode()
    return modes.iloc[0] if not modes.empty else np.nan

group = store.groupby("Region")["Sales"].agg(
    median="median",
    mode=first_mode,
    mean="mean",
    sum="sum"
)

print(group)
```

# 6. group on region and category wise

```python
group = store.groupby(["Region","Category"])

print(group)
```

# 7. Find region and category wise average sales 

```python
group = store.groupby(["Region","Category"])["Sales"].mean()

print(group)
```

# 8. Find region and category wise average, minimumm, maximum and sum sales 

```python
group = store.groupby(["Region","Category"])["Sales"].agg(["mean", "min", "max", "sum"])
```

# Merging and joining

```python
students = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110,
                   111, 112, 113, 114, 115, 116, 117, 118, 119, 120],
    "name": ["Rahim", "Karim", "Hasan", "Sakib", "Nabil",
             "Tanvir", "Fahim", "Rafi", "Imran", "Shakil",
             "Arif", "Jahid", "Mehedi", "Siam", "Tareq",
             "Rakib", "Adnan", "Nayeem", "Foysal", "Rony"],
    "department": ["CSE", "EEE", "CSE", "BBA", "CSE",
                   "EEE", "CSE", "BBA", "CSE", "EEE",
                   "CSE", "BBA", "EEE", "CSE", "BBA",
                   "CSE", "EEE", "CSE", "BBA", "EEE"]
})

students
```

```python
results = pd.DataFrame({
    "student_id": [101, 102, 103, 105, 106, 108, 109, 111, 112, 114,
                   121, 122, 123, 124, 125, 126, 127, 128, 129, 130],
    "math": [85, 72, 91, 68, 77, 88, 74, 95, 81, 69,
             78, 84, 92, 73, 89, 67, 80, 94, 76, 83],
    "python": [90, 75, 88, 72, 80, 92, 79, 96, 85, 71,
               82, 87, 95, 76, 91, 70, 84, 93, 78, 86]
})

results
```

```python
departments = pd.DataFrame({
    "dept_id": [1, 2, 3, 4],
    "department": [
        "CSE",
        "EEE",
        "BBA",
        "Civil"
    ],
    "faculty": [
        "Engineering",
        "Engineering",
        "Business",
        "Engineering"
    ]
})

departments
```

# 1. Create a dataframe contains students name,faculty and number on math and python

```python
merged = pd.merge(
    students,
    results,
    on="student_id"
)

merged = pd.merge(
    merged,
    departments,
    on="department"
)

final = merged[["name", "faculty", "math", "python"]]

print(final)
```

# 2. Department wise students count

```python
group = students["department"].value_counts()

print(group)
```

# 3. Faculty wise department count

```python
group = departments.groupby("faculty")["department"].count()

print(group)
```

# 4. Find department wise highest number on each subject

```python
merged = pd.merge(
    students,
    results,
    on="student_id"
)

result = merged.groupby("department")[["math", "python"]].max()

print(result)
```

# 5. Find faculty wise highest number on each subject

```python
merged = pd.merge(
    students,
    results,
    on="student_id"
)

merged = pd.merge(
    merged,
    departments,
    on="department"
)

result = merged.groupby("faculty")[["math", "python"]].max()

print(result)
```
