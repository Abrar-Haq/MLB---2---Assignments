# change data type and memory size of any np-array using astype()

```python
arr = np.array([1, 2, 3, 4, 5])

print("Original array:")
print(arr)
print("Original data type:", arr.dtype)
print("Original item size:", arr.itemsize, "bytes")


float_arr = arr.astype(float)

print("\nFloat array:")
print(float_arr)
print("New data type:", float_arr.dtype)
print("New item size:", float_arr.itemsize, "bytes")


int16_arr = arr.astype(np.int16)

print("\nInt16 array:")
print(int16_arr)
print("New data type:", int16_arr.dtype)
print("New item size:", int16_arr.itemsize, "bytes")
```

# stacking and spliting

```python
d = np.arange(27).reshape(3,9)
e = np.arange(28,55).reshape(3,9)
```

# use hstack() and vstack()

```python
print("Array D:")
print(d)

print("\nArray E:")
print(e)

horizontal = np.hstack((d, e))

print("\nHorizontal Stack:")
print(horizontal)

vertical = np.vstack((d, e))

print("\nVertical Stack:")
print(vertical)
```

# use hsplit() and vsplit()

```python
print("Array D:")
print(d)

print("\nHorizontal Split of D into 3 parts:")
h_split = np.hsplit(d, 3)

for i, part in enumerate(h_split):
    print(f"\nPart {i + 1}:")
    print(part)

vertical_array = np.arange(24).reshape(6, 4)

print("\nVertical Array:")
print(vertical_array)

print("\nVertical Split into 2 parts:")
v_split = np.vsplit(vertical_array, 2)

for i, part in enumerate(v_split):
    print(f"\nPart {i + 1}:")
    print(part)
```
